```python
import pandas as pd

parts_df = pd.read_csv("./data/parts_manual_labels_v1.csv")
```

```python
parts_df
```

```python
parts_df["Parts"].value_counts()
```

```python
# Get all unique part descriptions from P1-P5 columns
all_parts = []

# Iterate through P1-P5 columns
for col in ['P1', 'P2', 'P3', 'P4', 'P5']:
    # Get all non-null values from this column
    values = parts_df[col].dropna()
    
    # Split by comma and collect all parts
    for val in values:
        if pd.notna(val) and str(val).strip():  # Check if value exists
            # Split by comma
            parts = [p.strip() for p in str(val).split(',')]
            all_parts.extend(parts)

# Get unique values (case-sensitive) and sort
unique_parts = sorted(set(all_parts))

print(f"Total unique part descriptions: {len(unique_parts)}\n")
print("Unique part descriptions:")
print("-" * 50)
for part in unique_parts:
    print(f"  - {part}")
```

```python
online_catalogue_df = pd.read_csv("./data/ford-catalogue/online_catalogue.csv")
online_catalogue_df.head()

```

```python
# Add online catalogue info to parts_df
n_examples = []
clans_list = []
pods_list = []

for call in parts_df['Call']:
    # Filter matching rows
    matching_rows = online_catalogue_df[online_catalogue_df['call_type'] == call]
    
    # Count examples
    n_examples.append(len(matching_rows))
    
    # Get unique clans (comma-separated)
    unique_clans = matching_rows['clan'].dropna().unique()
    clans_list.append(', '.join(sorted(unique_clans)) if len(unique_clans) > 0 else '')
    
    # Get unique pods (comma-separated)
    unique_pods = matching_rows['pod'].dropna().unique()
    pods_list.append(', '.join(sorted(unique_pods)) if len(unique_pods) > 0 else '')

# Add columns to parts_df
parts_df['n_online_examples'] = n_examples
parts_df['clans'] = clans_list
parts_df['pods'] = pods_list

parts_df.head()
```

```python
parts_df.to_csv("./data/parts_manual_labels_v2.csv", index=False)
```

```python
import pandas as pd
pd.set_option('display.max_columns', None)
parts_v3 = pd.read_csv("./data/parts_manual_labels_v3.csv")
# parts_v3 = parts_v3[parts_v3["checked_v3"] == True]

parts_v3.reset_index(drop=True, inplace=True)
```

```python
parts_v3
```

```python
# Get all unique part descriptions from P1-P5 columns
all_parts = []

# Iterate through P1-P5 columns
for col in ['P1_desc', 'P2_desc', 'P3_desc', 'P4_desc', 'P5_desc']:
    # Get all non-null values from this column
    values = parts_v3[col].dropna()
    
    # Split by comma and collect all parts
    for val in values:
        if pd.notna(val) and str(val).strip():  # Check if value exists
            # Split by comma or |
            segs = []
            for chunk in str(val).split(','):
                segs.extend([p.strip() for p in chunk.split('|')])
            all_parts.extend([p for p in segs if p])

# Get unique values (case-sensitive) and sort
unique_parts = sorted(set(all_parts))

print(f"Total unique part descriptions: {len(unique_parts)}\n")
print("Unique part descriptions:")
print("-" * 50)
for part in unique_parts:
    print(f"  - {part}")
```

```python
import re
match_dict = {
    "single tone" : "SINGLE_TONE",
    "sbi increasing" : "SBI_INC",

    "sbi decreasing" : "SBI_DEC",
    "sbi narrowing" : "SBI_DEC",
    "tight" : "SBI_TIGHT",
    "wide" : "SBI_WIDE",
    "ascending" : "ASC",
    "descending" : "DESC",
    "upsweep" : "UPSWEEP",
    "downsweep" : "DOWNSWEEP",
    "squiggle" : "SQUIGGLE",
    "squigle" : "SQUIGGLE",
    "fast" : "FAST",
    "slow" : "SLOW",
    "peak" : "PEAK",
    "flat" : "FLAT",
    "biphonation" : "BIPHO",
    ",": ",",
    "|" : "|",
    "gap" : "GAP",
    "asending" : "ASC",
    "left" : "LEFT",
    "right" : "RIGHT",
    "dip" : "DIP",
    "large" : "LARGE",
    "small" : "SMALL",
    "rigth" : "RIGHT",
    "?" : "?",
    "flat?" : "FLAT ?",
    
    
    
}

parts_v3["P1_tokens"] = ""
parts_v3["P2_tokens"] = ""
parts_v3["P3_tokens"] = ""
parts_v3["P4_tokens"] = ""
parts_v3["P5_tokens"] = ""
parts_v3["tokens_string"] = ""

for i, call in parts_v3.iterrows():
    print(call["Call"])
    print(i)
    if call["checked_v3"] == False:
        print("skipping")
        continue

    main_string = ""

    token_string = ""
    for j in range(1, int(call["Parts"]) +1):
        
        part_desc = call[f"P{j}_desc"]
        part_type = call[f"P{j}_type"]
        part_length = "." + call[f"P{j}_length"] if call[f"P{j}_length"] != "undefined" else ""
        part_always_present = "!" if call[f"P{j}_always_present"] else ""
        part_variation = "~" if call[f"P{j}_strong_variation"] else ""
        
        if pd.notna(part_desc) and str(part_desc).strip():

            main_string += f"[{part_type}{part_length}{part_always_present}{part_variation}] {part_desc} "

        part_type = call[f"P{j}_type"]
        
        type_token = part_type.upper() if part_type.lower() != "tonal" else ""

        part_length = call[f"P{j}_length"] 
      
        length_token = part_length.upper() if part_length != "undefined" else ""

        # Tokenize the description
        part_desc = call[f"P{j}_desc"]
        desc_lower = str(part_desc).lower() if pd.notna(part_desc) else ""
        original_desc = desc_lower  # Keep original for comparison

        # Track matched positions
        matched_positions = set()

        # Sort by length (longest first) to match multi-word phrases first
        sorted_patterns = sorted(match_dict.items(), key=lambda x: len(x[0]), reverse=True)

        # Find all occurrences
        matches = []  # List of (position, pattern, token)
        for pattern, token in sorted_patterns:
            start = 0
            while True:
                pos = desc_lower.find(pattern, start)
                if pos == -1:
                    break
                # Check if this position overlaps with already matched text
                pattern_positions = set(range(pos, pos + len(pattern)))
                if not pattern_positions.intersection(matched_positions):
                    matches.append((pos, pattern, token))
                    matched_positions.update(pattern_positions)
                start = pos + 1

        # Sort matches by position to preserve order
        matches.sort(key=lambda x: x[0])
        desc_tokens = [match[2] for match in matches]

        # Build matched string for comparison
        desc_token_string = " ".join(desc_tokens) if desc_tokens else ""

        # Find unmatched text
        unmatched_chars = []
        for char_idx, char in enumerate(original_desc):
            if char_idx not in matched_positions:
                unmatched_chars.append(char)

        unmatched_text = ''.join(unmatched_chars).strip()
        # Clean up extra spaces and commas
        unmatched_text = ' '.join(unmatched_text.split())
        unmatched_text = unmatched_text.replace(' , ', ', ').strip(' ,')

        # print(f"P{j} original: {part_desc}")
        # print(f"P{j} tokens: {desc_token_string}")
        if unmatched_text:
            print(f"P{j} unmatched: {unmatched_text}")
        
        part_token_string = f"{type_token} {length_token} {desc_token_string}"
        part_token_string = part_token_string.rstrip()

        parts_v3.at[i, f"P{j}_tokens"] = part_token_string

        token_string += f"({part_token_string}) "
  
    # Remove any spaces after '(' or before ')' using regex
    token_string_clean = re.sub(r'\(\s+', '(', token_string)
    token_string_clean = re.sub(r'\s+\)', ')', token_string_clean)
    parts_v3.at[i, "tokens_string"] = token_string_clean

        
    print(main_string)
    print(token_string_clean)
```
