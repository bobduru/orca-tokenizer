```python
import pandas as pd


parts_df = pd.read_csv("./data/parts_manual_labels_v2.csv")

import numpy as np
import re

for i in range(1, 6):
    pi_col = f"P{i}"
    parts_df[f"{pi_col}_old"] = parts_df[pi_col]
    # Type: check for tonal/bb in any case, prioritizing tonal, then bb, else nan
    def get_type(val):
        val_str = str(val).lower()
        if "tonal" in val_str:
            return "tonal"
        elif "bb" in val_str:
            return "BB"
        return np.nan

    # Desc: remove 'BB', 'Tonal', 'tonal' at the start, rest as minuscule
    def get_desc(val):
        if pd.isna(val):
            return ""
        val_str = str(val).strip()
        # Remove leading 'BB', 'Tonal', or 'tonal', and any commas/spaces directly after
        desc = re.sub(r"^(BB|Tonal|tonal)[, ]*", "", val_str)
        return desc.lower().strip()

    parts_df[f"{pi_col}_type"] = parts_df[pi_col].apply(get_type)
    parts_df[f"{pi_col}_desc"] = parts_df[pi_col].apply(get_desc)
    parts_df[f"{pi_col}_length"] = "undefined"
    parts_df[f"{pi_col}_always_present"] = True
    parts_df[f"{pi_col}_strong_variation"] = False


parts_df["old_notes"] = parts_df["Notes"]
parts_df["checked_v3"] = False
```
