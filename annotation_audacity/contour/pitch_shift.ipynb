```python
import librosa
from utils import  plot_spectrogram
%load_ext autoreload
%autoreload 2

folder_path = "../../data/ford-catalogue/Northern Resident/media/A_clan"
# filename = "N01ii-I2-1"
filename = "N01iii-C1-1"
filename = "N03-A1-1"

# Load audio
y, sr = librosa.load(f"{folder_path}/{filename}.mp3")

n_fft = 2048    
hop_length = 256
plot_spectrogram(y, sr, title="Spectrogram: original audio", figsize=(7,4), n_fft=n_fft, hop_length=hop_length)

shifted_audios = []
for i in range(1, 4):
    print(f"Pitch shift: +{i} octaves")
    # Pitch shift: +i octaves up (12 semitones per octave)
    y_shifted = librosa.effects.pitch_shift(y, sr=sr, n_steps=i*24)
    shifted_audios.append(y_shifted)
    # Plot spectrogram of shifted audio (in-memory, no file needed)
    plot_spectrogram(y_shifted, sr, title=f"Spectrogram: pitch-shifted +{i} semitones", figsize=(7,4), n_fft=n_fft, hop_length=hop_length)

```

```python
import IPython.display as ipd

# Play the 3rd shifted audio (Python is 0-indexed, so index 2)
ipd.display(ipd.Audio(shifted_audios[0], rate=sr))
```
