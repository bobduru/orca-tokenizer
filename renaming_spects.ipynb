```python
import pandas as pd

parts_df = pd.read_csv("./data/parts_manual_labels_v1.csv")
```

```python
parts_df
```

```python
import os
from pathlib import Path

# Get the directory with screenshots
image_dir = Path("./data/ford_paper_spects/")

# Get all screenshot files and sort by modification time (oldest first)
screenshot_files = sorted(
    image_dir.glob("*.png"),
    key=lambda x: x.stat().st_mtime
)

print(f"Found {len(screenshot_files)} screenshot files")
print(f"Found {len(parts_df)} calls in parts_df")

# Verify counts match
if len(screenshot_files) == len(parts_df):
    print("\n✓ Counts match! Ready to rename.")
else:
    print(f"\n⚠ Warning: {len(screenshot_files)} files but {len(parts_df)} calls")
    
```

```python
# Perform the renaming
renamed_count = 0
for i, old_file in enumerate(screenshot_files):
    call_name = parts_df.iloc[i]['Call']
    new_name = f"{call_name}_paper_spect.png"
    new_path = old_file.parent / new_name
    
    # Rename the file
    old_file.rename(new_path)
    renamed_count += 1
    
    # Print progress every 10 files
    if (i + 1) % 10 == 0:
        print(f"Renamed {i + 1}/{len(screenshot_files)} files...")

print(f"\n✓ Successfully renamed {renamed_count} files!")
print("\nSample of renamed files:")
renamed_files = sorted(image_dir.glob("*_paper_spect.png"))[:5]
for f in renamed_files:
    print(f"  - {f.name}")
```
